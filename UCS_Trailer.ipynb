### Uniform Cost Search Algorithm

#### Define a dictionary

```python
# Make a dictionary with each city and its neighbouring nodes
cost_graph = {
  "Nairobi": [("Nakuru", 160), ("Narok", 141), ("Thika", 44)],
  "Thika": [("Murang'a", 46), ("Nyeri", 108)],
  "Murang'a": [("Nyeri", 54)],
  "Nyeri": [("Nakuru", 165), ("Nanyuki", 59)],
  "Nanyuki": [("Nakuru", 158)],
  "Nakuru": [("Nairobi", 160), ("Nyeri", 165), ("Narok", 123), ("Kericho", 107)],
  "Narok": [("Nairobi", 141)],
  "Kericho": [("Nakuru", 107), ("Bomet", 76), ("Kisumu", 83), ("Kisii", 90)],
  "Bomet": [("Narok", 82)],
  "Kisii": [("Bomet", 84)],
  "Kisumu": [("Kisii", 112)]
}
```

#### Add import statement

```python
import heapq
```

#### Search Function

```python
def ucs(graph, start, goal):
    # Sets the queue to begin at the start point with a distance of 0
    queue = [(0, [start])]
    # Create an empty set called visited
    visited = set()

    while queue:
        # Gets the node from the queue, cost, path and current city in the list
        current = heapq.heappop(queue)
        current_cost = current[0]
        current_path = current[1]
        current_city = current_path[-1]

        # End the loop and return the path and cost if the current city is the goal
        if current_city == goal:
            return [current_path, current_cost]

        if current_city not in visited:
            # Adds city to the set if it has not been explored before
            visited.add(current_city)

            # Loop through each neighbouring node from the graph
            for location in graph.get(current_city, []):
                neighbouring_city = location[0]
                path_cost = location[1]

                # Finds the new total cost for the path and adds the city to the path if it has not been visited
                if neighbouring_city not in visited:
                    new_cost = current_cost + path_cost
                    new_path = current_path + [neighbouring_city]

                    # Adds the path to the queue
                    heapq.heappush(queue, (new_cost, new_path))

        print(queue)

    return [None, -1]
```

#### Running the UCS algorithm

```python
starting_location = "Nairobi"
destination = "Kisumu"

# Call the function and get the path and cost values
result = ucs(cost_graph, starting_location, destination)
path = result[0]
total_distance = result[1]

output_path = ""

# Loop through the path to write values to the output string
for i in range(len(path)):
    city = path[i]
    if i == 0:
        output_path = city
    else:
        output_path = output_path + " - " + city

print("\n")

print(f"Best Path: {output_path}")
print(f"Total Distance: {total_distance} km")
```
